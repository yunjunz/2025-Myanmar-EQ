### Save the RdBu colormap into GMT-compatible CPT file

```python
%matplotlib inline
import os
import matplotlib as mpl
from matplotlib import pyplot as plt
from mintpy.objects.colors import read_cpt_file, save_cpt_file

work_dir = os.path.expanduser('~/data/archives/2025MyanmarEQ/notebooks/data')
os.chdir(work_dir)
print('Go to directory:', work_dir)

cpt_file = os.path.expanduser('~/data/archives/2025MyanmarEQ/notebooks/data/RdBu.cpt')
```

```python
# save the matplotlib colormap into CPT file
save_cpt_file('RdBu', cpt_file=cpt_file)
```

```python
# read
cmap1 = read_cpt_file(cpt_file, cmap_lut=256)
cmap2 = read_cpt_file(cpt_file, cmap_lut=8)

# plot
fig, axs = plt.subplots(nrows=2, ncols=1, figsize=[5, 0.5], sharex=True, sharey=True)
mpl.colorbar.ColorbarBase(axs[0], cmap=cmap1, orientation='horizontal')
mpl.colorbar.ColorbarBase(axs[1], cmap=cmap2, orientation='horizontal')

# axis format
for ax in axs:
    ax.set_xticklabels([])
fig.subplots_adjust(hspace=0)
axs[0].set_title('RdBu', fontsize=12, weight='bold')
```
