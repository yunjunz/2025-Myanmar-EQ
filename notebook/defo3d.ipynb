```python
atr = readfile.read_attribute(dis_files[0])
length, width = int(atr['LENGTH']), int(atr['WIDTH'])
# calculate reference point in y/x

dis     = np.full((num_file, length, width), np.nan, dtype=np.float32)
dis_std = np.full((num_file, length, width), np.nan, dtype=np.float32)
ve      = np.full((num_file, length, width), np.nan, dtype=np.float32)
vn      = np.full((num_file, length, width), np.nan, dtype=np.float32)
vu      = np.full((num_file, length, width), np.nan, dtype=np.float32)

for i, (dis_file, dis_std_file, enu_file) in enumerate(zip(dis_files, dis_std_files, enu_files)):
    # read displacement
    print(f'read data from file {i+1}: {dis_file}')
    dis_i, atr_i = readfile.read(dis_file)
    # convert phase to range
    if dis_file.endswith('.unw'):
        phase2range = float(atr_i['WAVELENGTH']) / (4.0 * np.pi)
        dis_i *= phase2range

    # read displacement std
    if isinstance(dis_std_file, str):
        dis_std_i = readfile.read(dis_std_file)[0]
    else:
        dis_std_i = np.ones(dis_i.shape, dtype=np.float32) * dis_std_file

    # read unit vector
    ve_i = readfile.read(enu_file, datasetName='band1')[0]
    vn_i = readfile.read(enu_file, datasetName='band2')[0]
    vu_i = readfile.read(enu_file, datasetName='band3')[0]

    # spatial referencing
    if dis_file.endswith('.unw'):
        coord = ut.coordinate(atr_i)
        ref_lat_al , ref_lon_al = 22,96.6
        ref_y, ref_x = coord.geo2radar(ref_lat_al, ref_lon_al)[:2]
        print(f'\tspatial referencing to (lat, lon) = ({ref_lat}, {ref_lon}), (y, x) = ({ref_y}, {ref_x})')
        if np.isnan(dis_i[ref_y, ref_x]):
            raise ValueError('reference point is in masked out region! Choose a different one!')
        dis_i -= dis_i[ref_y, ref_x]

    # resize
    if dis_i.shape != (length, width):
        print(f'\tresizing shape from {dis_i.shape} to {(length,width)}')
        kwargs = dict(output_shape=(length, width), order=0, mode='constant', anti_aliasing=False, preserve_range=True)
        #kwargs = dict(output_shape=(length, width), order=1, mode='constant', anti_aliasing=True, preserve_range=True)
        dis_i     = resize(np.array(dis_i),     **kwargs)
        dis_std_i = resize(np.array(dis_std_i), **kwargs)
        ve_i      = resize(np.array(ve_i),      **kwargs)
        vn_i      = resize(np.array(vn_i),      **kwargs)
        vu_i      = resize(np.array(vu_i),      **kwargs)

    dis[i] = dis_i
    dis_std[i] = dis_std_i
    ve[i] = ve_i
    vn[i] = vn_i
    vu[i] = vu_i
print('finished reading data.')
```

## 3. Estimate 3D displacement field

### 3.1 Estimate 3D displacement

```python
def estimate_enu_displacement(dis_los, dis_los_std, G, weight=False):

    dis_los = dis_los.reshape(G.shape[0], -1)
    dis_los_std = dis_los_std.reshape(G.shape[0], -1)

    # initial output value
    dis_enu     = np.full(3, np.nan, dtype=np.float32)
    dis_enu_std = np.full(3, np.nan, dtype=np.float32)

    # skip pixels with none in all files
    if np.all(np.isnan(dis_los)):
        return dis_enu, dis_enu_std

    # skip invalid observations in some files
    dis_los, [G, dis_los_std] = skip_invalid_obs(dis_los, mat_list=[G, dis_los_std])

    # invert
    if weight:
        w_sqrt = 1. / dis_los_std
        X, e2 = linalg.lstsq(np.multiply(G, w_sqrt), np.multiply(dis_los, w_sqrt), cond=1e-5)[:2]
    else:
        X, e2 = linalg.lstsq(G, dis_los, cond=1e-5)[:2]

    if e2.size != 0:
        dis_enu = X.flatten()

        # calculate inversion quality
        Gplus = linalg.pinv(G)
        dis_los_cov = np.diag(np.square(dis_los_std.flatten()))
        dis_enu_cov = np.linalg.multi_dot([Gplus, dis_los_cov, Gplus.T])
        dis_enu_std = np.sqrt(np.diag(dis_enu_cov))

    return dis_enu, dis_enu_std
```

```python
# calculate the weight matrix
weight = True

# pixel-wise inversion with invalid pixels skipped.
dis_enu = np.zeros((3, length, width), dtype=np.float32) * np.nan
dis_enu_std = np.zeros((3, length, width), dtype=np.float32) * np.nan
num_pix = length * width
prog_bar = ptime.progressBar(maxValue=num_pix)
for i in range(num_pix):
    prog_bar.update(i+1, every=1000, suffix=f'{i+1} / {num_pix}')
    y, x = i // width, i % width

    # prepare inputs
    dis_los = dis[:, y, x]
    dis_los_std = dis_std[:, y, x]
    G = np.zeros((num_file, 3), dtype=np.float32)
    G[:, 0] = ve[:, y, x]
    G[:, 1] = vn[:, y, x]
    G[:, 2] = vu[:, y, x]

    dis_enu_i, dis_enu_std_i = estimate_enu_displacement(dis_los, dis_los_std, G, weight=weight)
    dis_enu[:, y, x] = dis_enu_i.flatten()
    dis_enu_std[:, y, x] = dis_enu_std_i.flatten()
prog_bar.close()
```

### 3.2 Write result to HDF5 file

```python
## mask
# 1. mask out pixels with large STD
# 2. mask out pixels with abnormal values
max_std_list = [0.5, 1.5, 0.5]
max_dis_list = [8, 10, 8]
print(f'mask out pixels with large STD: east > {max_std_list[0]} m, north > {max_std_list[1]} m, up > {max_std_list[2]} m.')
print(f'mask out pixels with large val: east > {max_dis_list[0]} m, north > {max_dis_list[1]} m, up > {max_dis_list[2]} m.')
for i, (max_std, max_dis) in enumerate(zip(max_std_list, max_dis_list)):
    dis_enu[i][dis_enu_std[i] > max_std] = np.nan
    dis_enu[i][dis_enu[i] > max_dis] = np.nan
# 3. mask out sparse pixels in the far field (for a clean picture)
# the maskAOI.h5 is generated using generate_mask.py --roipoly option
mask_aoi_file = os.path.join(work_dir, 'defo3d/maskAOI.h5')
if os.path.isfile(mask_aoi_file):
    print(f'mask out pixels marked in file: {mask_aoi_file}')
    mask_aoi = readfile.read(mask_aoi_file)[0]
    for i in range(3):
        dis_enu[i][mask_aoi==1] = np.nan

## spatial reference (to a point in the far-field)
# instead of a common pixels used above for stitching purpose only
ref_lat, ref_lon =  20 ,94.6
ref_y, ref_x = ut.coordinate(atr).geo2radar(ref_lat, ref_lon)[:2]
print(f'spatial referencing to point: lat / lon = {ref_lat} / {ref_lon}, y / x = {ref_y} / {ref_x}.')
for i in range(3):
    dis_enu[i] -= dis_enu[i][ref_y, ref_x]

## write to HDF5 file
for data, fname in zip([dis_enu, dis_enu_std], [dis_enu_file, dis_enu_std_file]):
    ds_dict = {
        'east'  : data[0].reshape(length, width),
        'north' : data[1].reshape(length, width),
        'up'    : data[2].reshape(length, width),
    }
    atr['FILE_TYPE'] = 'displacement'
    atr['DATA_TYPE'] = 'float32'
    atr['UNIT'] = 'm'
    atr['BANDS'] = len(ds_dict.keys())
    atr['REF_LAT'], atr['REF_LON'] = ref_lat, ref_lon
    atr['REF_Y'], atr['REF_X'] = ref_y, ref_x
    writefile.write(ds_dict, fname, metadata=atr)
```

```python
# prep displacement up
cmd = f'view.py {dis_enu_file} up --dem {dem_file} --dem-nocontour --shade-exag 0.05 --faultline {flt_file} '
cmd += f' -c RdBu -v -0.5 0.5 -u m --lalo-label --lalo-step 2 --ylabel-rot 90 --noverbose --notitle --fontsize 12 --nocbar '
cmd += f' --sub-lat {S} {N} --sub-lon {W} {E} --scalebar 0.3 0.8 0.05 '
cmd += f' --ref-size 4 '
dis_u, atr, inps = prep_slice(cmd)

# plot horizontal displacement
fig, ax = plt.subplots(figsize=[6, 6], subplot_kw=dict(projection=ccrs.PlateCarree()))
ax, inps, im, cbar = plot_slice(ax, dis_u, atr, inps)

# ... existing code ...

def generate_quiver_grid_by_fault_distance_simple(flt_file, dis_e, dis_n, lons, lats, atr, 
                                                lookup_file, distance_step=30, 
                                                max_distance=300, window_radius=2):
    """
    重新设计：先找到所有等值线上的点，再采样控制密度
    """
    from mintpy.utils import readfile
    from mintpy.utils import utils as ut
    from scipy.spatial import KDTree
    import numpy as np
    
    print(f'Reading fault file: {flt_file}')
    
    # 读取断层文件
    try:
        faults_lola = readfile.read_gmt_lonlat_file(flt_file, min_dist=0)
        print(f'Found {len(faults_lola)} fault segments')
    except Exception as e:
        print(f'Error reading fault file: {e}')
        return [], [], [], []
    
    # 转换断层到雷达坐标系
    coord = ut.coordinate(atr, lookup_file=lookup_file)
    fault_points = []
    
    for fault_lola in faults_lola[0]:
        # 转换断层点到雷达坐标系
        y, x = coord.geo2radar(fault_lola[1], fault_lola[0])[:2]
        fault_points.append([x, y])
    
    if not fault_points:
        print('No valid fault points found')
        return [], [], [], []
    
    # 创建KDTree用于快速距离计算
    fault_tree = KDTree(fault_points)
    print(f'Created KDTree with {len(fault_points)} fault points')
    
    # 生成非线性箭头网格
    print(f'Generating nonlinear quiver grid with distance step {distance_step} pixels...')
    
    # 存储所有找到的点
    all_points = []  # [(lon, lat, e, n, distance), ...]
    
    length, width = dis_e.shape
    
    # 第一步：找到所有距离等值线上的点
    for target_distance in np.arange(distance_step, max_distance + distance_step, distance_step):
        print(f'Processing distance contour at {target_distance} pixels...')
        
        tolerance = distance_step / 2  # 允许的误差范围
        contour_points = []  # 当前等值线上的点
        
        # 在整个图像中寻找距离接近目标距离的点
        for i in range(0, length, 5):  # 每5个像素采样一次，提高精度
            for j in range(0, width, 5):
                dist, _ = fault_tree.query([j, i])
                
                # 检查是否在目标距离范围内
                if abs(dist - target_distance) <= tolerance:
                    # 定义采样窗口
                    y_start = max(0, i - window_radius)
                    y_end = min(length, i + window_radius + 1)
                    x_start = max(0, j - window_radius)
                    x_end = min(width, j + window_radius + 1)
                    
                    # 提取窗口内的数据
                    window_e = dis_e[y_start:y_end, x_start:x_end]
                    window_n = dis_n[y_start:y_end, x_start:x_end]
                    
                    # 计算平均值（忽略NaN值）
                    if not np.all(np.isnan(window_e)) and not np.all(np.isnan(window_n)):
                        avg_e = np.nanmean(window_e)
                        avg_n = np.nanmean(window_n)
                        
                        # 只有当平均值不是NaN时才添加点
                        if not (np.isnan(avg_e) or np.isnan(avg_n)):
                            contour_points.append((lons[j], lats[i], avg_e, avg_n, dist))
        
        print(f'  Found {len(contour_points)} points at distance {target_distance}')
        all_points.extend(contour_points)
    
    print(f'Total points found: {len(all_points)}')
    
    # 第二步：对找到的点进行采样，控制最终密度
    if len(all_points) > 0:
        # 按距离分组
        distance_groups = {}
        for point in all_points:
            dist = point[4]
            dist_key = int(dist / distance_step) * distance_step
            if dist_key not in distance_groups:
                distance_groups[dist_key] = []
            distance_groups[dist_key].append(point)
        
        # 对每个距离组进行采样
        arrow_lons = []
        arrow_lats = []
        arrow_e = []
        arrow_n = []
        
        for dist_key, points in distance_groups.items():
            # 增加每个距离组的采样点数
            target_points_per_group = min(500, len(points))  # 每个组最多20个点，或者所有可用点
            
            if len(points) <= target_points_per_group:
                selected_points = points
            else:
                # 按经度排序，确保断层两侧都有点
                points.sort(key=lambda x: x[0])
                left_points = [p for p in points if p[0] < np.mean([fault[0] for fault in faults_lola[0]])]
                right_points = [p for p in points if p[0] >= np.mean([fault[0] for fault in faults_lola[0]])]
                
                # 从左右两侧按比例取点
                selected_points = []
                if len(left_points) > 0:
                    left_target = max(1, target_points_per_group // 2)
                    step = max(1, len(left_points) // left_target)
                    selected_points.extend(left_points[::step][:left_target])
                
                if len(right_points) > 0:
                    right_target = target_points_per_group - len(selected_points)
                    step = max(1, len(right_points) // right_target)
                    selected_points.extend(right_points[::step][:right_target])
                
                # 如果点数还不够，从剩余点中补充
                remaining_points = [p for p in points if p not in selected_points]
                if len(selected_points) < target_points_per_group and len(remaining_points) > 0:
                    step = max(1, len(remaining_points) // (target_points_per_group - len(selected_points)))
                    selected_points.extend(remaining_points[::step][:(target_points_per_group - len(selected_points))])
            
            # 添加到最终结果
            for point in selected_points:
                arrow_lons.append(point[0])
                arrow_lats.append(point[1])
                arrow_e.append(point[2])
                arrow_n.append(point[3])
        
        print(f'After sampling: {len(arrow_e)} arrows')
        
        # 检查采样点的空间分布
        lons_array = np.array(arrow_lons)
        lats_array = np.array(arrow_lats)
        print(f'Sampling points longitude range: {lons_array.min():.3f} to {lons_array.max():.3f}')
        print(f'Sampling points latitude range: {lats_array.min():.3f} to {lats_array.max():.3f}')
        
        # 检查断层两侧的分布
        fault_lon = np.mean([fault[0] for fault in faults_lola[0]])
        left_points = sum(1 for lon in arrow_lons if lon < fault_lon)
        right_points = sum(1 for lon in arrow_lons if lon >= fault_lon)
        print(f'Points left of fault: {left_points}, Points right of fault: {right_points}')
        
        return arrow_lons, arrow_lats, arrow_e, arrow_n
    else:
        return [], [], [], []


# 使用简化版本
arrow_lons, arrow_lats, arrow_e, arrow_n = generate_quiver_grid_by_fault_distance_simple(
    flt_file=flt_file,
    dis_e=dis_e,
    dis_n=dis_n,
    lons=lons,
    lats=lats,
    atr=atr,
    lookup_file=lookup_file,
    distance_step=50,    # 每30像素生成一个采样点
    max_distance=300,    # 最大距离300像素
    window_radius=2      # 采样窗口半径2像素
)


# 绘制方向箭头
Q = ax.quiver(arrow_lons, arrow_lats, arrow_e, arrow_n,
              headwidth=1.5, alpha=0.7)

# axis format
cax = fig.add_axes([0.35, 0.17, 0.12, 0.01])
cbar = fig.colorbar(im, cax=cax, orientation='horizontal', ticks=[-0.5, 0, 0.5])
cbar.ax.xaxis.set_label_position('top')
cbar.ax.set_xlabel('vert [m]', fontsize=12)
ax.quiverkey(Q, X=0.53, Y=0.14, U=1, label='1 m\nhorz', labelsep=0.08, coordinates='figure', alpha=0.8, linewidth=30)

# output
out_fig = os.path.join(work_dir, 'defo3d/pic/dis_u_new_test.png')
print('save figure to file', out_fig)
plt.savefig(out_fig, bbox_inches='tight', transparent=True, dpi=600)
plt.show()
```

```python
# prep displacement up
cmd = f'view.py {dis_enu_file} up --dem {dem_file} --dem-nocontour --shade-exag 0.05 --faultline {flt_file} '
cmd += f' -c RdBu -v -0.5 0.5 -u m --lalo-label --lalo-step 2 --ylabel-rot 90 --noverbose --notitle --fontsize 12 --nocbar '
cmd += f' --sub-lat {S} {N} --sub-lon {W} {E} --scalebar 0.3 0.8 0.05 '
cmd += f' --ref-size 4 '
dis_u, atr, inps = prep_slice(cmd)

# plot horizontal displacement
fig, ax = plt.subplots(figsize=[6, 6], subplot_kw=dict(projection=ccrs.PlateCarree()))
ax, inps, im, cbar = plot_slice(ax, dis_u, atr, inps)

# ... existing code ...

def generate_quiver_grid_by_fault_distance_kdtree(flt_file, dis_e, dis_n, lons, lats, atr, 
                                                lookup_file, min_distance=5, max_distance=50, 
                                                grid_spacing=20, window_radius=2):
    """
    使用KDTree加速的距离计算版本
    """
    from mintpy.utils import readfile
    from mintpy.utils import utils as ut
    from scipy.spatial import KDTree
    from scipy.interpolate import griddata
    
    print(f'Reading fault file: {flt_file}')
    
    # 读取断层文件
    try:
        faults_lola = readfile.read_gmt_lonlat_file(flt_file, min_dist=0)
        print(f'Found {len(faults_lola)} fault segments')
    except Exception as e:
        print(f'Error reading fault file: {e}')
        return [], [], [], []
    
    # 转换断层到雷达坐标系
    coord = ut.coordinate(atr, lookup_file=lookup_file)
    fault_points = []
    
    for fault_lola in faults_lola[0]:
        # 转换断层点到雷达坐标系
        y, x = coord.geo2radar(fault_lola[1], fault_lola[0])[:2]
        fault_points.append([x, y])
    
    if not fault_points:
        print('No valid fault points found')
        return [], [], [], []
    
    # 创建KDTree用于快速距离计算
    fault_tree = KDTree(fault_points)
    print(f'Created KDTree with {len(fault_points)} fault points')
    
    # 生成箭头网格
    print('Generating quiver grid based on fault distance...')
    
    arrow_lons = []
    arrow_lats = []
    arrow_e = []
    arrow_n = []
    
    length, width = dis_e.shape
    
    # 在指定距离范围内生成网格点
    for i in range(grid_spacing, length-grid_spacing, grid_spacing):
        for j in range(grid_spacing, width-grid_spacing, grid_spacing):
            # 计算到断层的最短距离
            dist, _ = fault_tree.query([j, i])
            
            # 检查是否在指定距离范围内
            if min_distance <= dist <= max_distance:
                # 定义采样窗口
                y_start = max(0, i - window_radius)
                y_end = min(length, i + window_radius + 1)
                x_start = max(0, j - window_radius)
                x_end = min(width, j + window_radius + 1)
                
                # 提取窗口内的数据
                window_e = dis_e[y_start:y_end, x_start:x_end]
                window_n = dis_n[y_start:y_end, x_start:x_end]
                
                # 计算平均值（忽略NaN值）
                if not np.all(np.isnan(window_e)) and not np.all(np.isnan(window_n)):
                    avg_e = np.nanmean(window_e)
                    avg_n = np.nanmean(window_n)
                    
                    # 只有当平均值不是NaN时才添加箭头
                    if not (np.isnan(avg_e) or np.isnan(avg_n)):
                        arrow_e.append(avg_e)
                        arrow_n.append(avg_n)
                        arrow_lons.append(lons[j])
                        arrow_lats.append(lats[i])
    
    print(f'Generated {len(arrow_e)} arrows in distance range {min_distance}-{max_distance} pixels')
    
    return arrow_lons, arrow_lats, arrow_e, arrow_n

# 使用KDTree版本
arrow_lons, arrow_lats, arrow_e, arrow_n = generate_quiver_grid_by_fault_distance_kdtree(
    flt_file=flt_file,
    dis_e=dis_e,
    dis_n=dis_n,
    lons=lons,
    lats=lats,
    atr=atr,
    lookup_file=lookup_file,
    min_distance=10,    # 距离断层至少10像素
    max_distance=200,   # 距离断层最多100像素
    grid_spacing=50,    # 网格间距30像素
    window_radius=2     # 采样窗口半径2像素
)

# 绘制方向箭头
Q = ax.quiver(arrow_lons, arrow_lats, arrow_e, arrow_n,
              headwidth=1.5, alpha=0.7)

# axis format
cax = fig.add_axes([0.35, 0.17, 0.12, 0.01])
cbar = fig.colorbar(im, cax=cax, orientation='horizontal', ticks=[-0.5, 0, 0.5])
cbar.ax.xaxis.set_label_position('top')
cbar.ax.set_xlabel('vert [m]', fontsize=12)
ax.quiverkey(Q, X=0.53, Y=0.14, U=1, label='1 m\nhorz', labelsep=0.08, coordinates='figure', alpha=0.8, linewidth=30)

# output
out_fig = os.path.join(work_dir, 'defo3d/pic/dis_u_new.png')
print('save figure to file', out_fig)
plt.savefig(out_fig, bbox_inches='tight', transparent=True, dpi=600)
plt.show()
```
